```python
class Nodo:
    def __init__(self, valor):
        self.valor = valor
        self.izq = None
        self.der = None
        self.altura = 1

class ArbolAVL:
    def obtener_altura(self, nodo):
        return nodo.altura if nodo else 0

    def factor_balance(self, nodo):
        return self.obtener_altura(nodo.izq) - self.obtener_altura(nodo.der) if nodo else 0

    def obtener_minimo(self, nodo):
        actual = nodo
        while actual.izq is not None:
            actual = actual.izq
        return actual

    # Rotaciones para el balanceo
    def rotar_derecha(self, y):
        x = y.izq
        T2 = x.der
        x.der = y
        y.izq = T2
        y.altura = 1 + max(self.obtener_altura(y.izq), self.obtener_altura(y.der))
        x.altura = 1 + max(self.obtener_altura(x.izq), self.obtener_altura(x.der))
        return x

    def rotar_izquierda(self, x):
        y = x.der
        T2 = y.izq
        y.izq = x
        x.der = T2
        x.altura = 1 + max(self.obtener_altura(x.izq), self.obtener_altura(x.der))
        y.altura = 1 + max(self.obtener_altura(y.izq), self.obtener_altura(y.der))
        return y

    # Método de eliminación y rebalanceo
    def eliminar(self, raiz, valor):
        # 1. Eliminación estándar BST
        if not raiz:
            return raiz

        if valor < raiz.valor:
            raiz.izq = self.eliminar(raiz.izq, valor)
        elif valor > raiz.valor:
            raiz.der = self.eliminar(raiz.der, valor)
        else:
            # Caso 1 y 2: Un solo hijo o sin hijos
            if raiz.izq is None:
                return raiz.der
            elif raiz.der is None:
                return raiz.izq

            # Caso 3: Dos hijos (sucesor in-order)
            temp = self.obtener_minimo(raiz.der)
            raiz.valor = temp.valor
            raiz.der = self.eliminar(raiz.der, temp.valor)

        if raiz is None:
            return raiz

        # 2. Actualizar altura
        raiz.altura = 1 + max(self.obtener_altura(raiz.izq), self.obtener_altura(raiz.der))

        # 3. Obtener factor de balance
        balance = self.factor_balance(raiz)

        # 4. Casos de desbalanceo
        # Izquierda - Izquierda
        if balance > 1 and self.factor_balance(raiz.izq) >= 0:
            return self.rotar_derecha(raiz)

        # Izquierda - Derecha
        if balance > 1 and self.factor_balance(raiz.izq) < 0:
            raiz.izq = self.rotar_izquierda(raiz.izq)
            return self.rotar_derecha(raiz)

        # Derecha - Derecha
        if balance < -1 and self.factor_balance(raiz.der) <= 0:
            return self.rotar_izquierda(raiz)

        # Derecha - Izquierda
        if balance < -1 and self.factor_balance(raiz.der) > 0:
            raiz.der = self.rotar_derecha(raiz.der)
            return self.rotar_izquierda(raiz)

        return raiz
```
